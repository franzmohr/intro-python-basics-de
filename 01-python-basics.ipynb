## Python Refresh

### Zahlen

```python
a = 5 # Teile der neuen Variable a den Wert 5 zu
b = 3 # Teile der neuen Variable b den Wert 3 zu
```

Eine Zuweisung mit `=` erzeugt keine Ausgabe. Die Werte sind jetzt aber in `a` und `b` gespeichert und können weiterverwendet werden.

```python
# Zeige den Wert der Variable a an
a
```

Steht ein Variablenname allein in der letzten Zeile einer Zelle, zeigt Jupyter ihren Wert an.

```python
a * b # Multipliziere a und b miteinander
```

```python
round(a / b, 4) # Dividiere a durch b und runde auf 4 Stellen genau
```

`a / b` ergibt 1,666…, `round(…, 4)` rundet auf vier Nachkommastellen. Python verwendet den Punkt als Dezimaltrennzeichen.

```python
a == b # Kontrolliere, ob a gleich b
```

```python
a < 5 # Kontrolliere, ob a kleiner 5
```

```python
a != b  # Kontrolliere, ob a ungleich b
```

Vergleiche liefern immer einen Wahrheitswert (*Boolean*): `True` oder `False`. `a == b` ist falsch, weil 5 nicht gleich 3 ist. `a < 5` ist falsch, weil 5 nicht *kleiner* als 5 ist (dafür bräuchte es `<=`). `a != b` ist wahr, weil sich die beiden Werte unterscheiden.

## Text

```python
c = 'data' # Definiere Variable c als 'data'
d = 'mining' # Definiere Variable d als 'mining'
e = c + d
e
```

Bei Text (*Strings*) bedeutet `+` nicht Addition, sondern Aneinanderhängen. Da kein Leerzeichen dazwischen steht, entsteht `'datamining'`. Die Anführungszeichen in der Ausgabe zeigen, dass es sich um einen String handelt.

```python
# Erstelle eine Liste welcome_list mit mehreren Wörtern
welcome_list = ['Welcome', 'to', 'Python', 'World.', 'It\'s', 'Amazing']
welcome_list
```

Eine Liste wird mit eckigen Klammern erstellt und kann beliebig viele Elemente enthalten. Das Apostroph in `It's` wird mit `\'` geschrieben, damit Python es nicht als Ende des Strings interpretiert. In der Ausgabe verwendet Python für diesen Eintrag deshalb doppelte Anführungszeichen.

```python
# Gib jedes Wort in welcome_list einzeln aus
for word in welcome_list:
    print(word)
```

Die `for`-Schleife geht die Liste Element für Element durch. `print` gibt jedes Wort in einer eigenen Zeile und ohne Anführungszeichen aus. Der eingerückte Block wird für jedes Element einmal ausgeführt.

```python
# Generiere einen Satz, indem die einzelnen Wörter in welcome_list
# mit einem Leerzeichen miteinander verbunden werden
welcome_sentence = ' '.join(welcome_list)
welcome_sentence
```

`' '.join(liste)` verbindet die Elemente einer Liste zu einem einzigen String und setzt dabei das Zeichen vor `.join` (hier ein Leerzeichen) zwischen die Elemente.

```python
# Extrahiere die ersten 10 Stellen des Satzes.
welcome_sliced = welcome_sentence[0:10]
welcome_sliced
```

Python zählt ab 0. `[0:10]` liefert die Zeichen an den Positionen 0 bis 9, die Position 10 ist nicht mehr enthalten. Leerzeichen zählen als Zeichen, deshalb ergibt sich `'Welcome to'`.

```python
# Transformiere alle Buchstaben in Großbuchstaben
welcome_sentence.upper()
```

`.upper()` ist eine *Methode* von Strings: eine Funktion, die direkt auf das Objekt angewendet wird. Der ursprüngliche Satz in `welcome_sentence` bleibt dabei unverändert, es wird ein neuer String zurückgegeben.

## Datenvisualisierung mit `matplotlib`

`matplotlib` (https://matplotlib.org/) ist ein äußerst beliebtes Paket für die Erstellung von Grafiken mit Python. Die Grafikfunktionalität des Pakets kann auf folgende Weise geladen werden:

```python
# Lade das Modul pyplot aus dem Paket matplotlib und mach es über die
# Abkürzung plt verfügbar
import matplotlib.pyplot as plt
import pandas
```

Ein `import` erzeugt keine Ausgabe. Ab jetzt sind die Funktionen von `matplotlib.pyplot` unter `plt` und die von `pandas` unter `pandas` verfügbar.

```python
# Lade einen Beispieldatensatz
pwt = pandas.read_excel("pwt100.xlsx",
                        sheet_name = "Data")

# Filter nach AT
pwt_at = pwt[pwt["countrycode"] == "AUT"]

# Wähle einzelne Spalten aus
pwt_at = pwt_at[["year", "countrycode", "rgdpe", "pop", "emp", "hc"]]

# Zeige Ergebnis an
pwt_at.head()
```

Die Penn World Table (PWT) enthält volkswirtschaftliche Kennzahlen für viele Länder. Wir behalten nur Österreich (`AUT`) und einige Spalten:

* `year`: Jahr
* `rgdpe`: reales BIP (ausgabenseitig) in Mio. US-Dollar zu Preisen von 2017
* `pop`: Bevölkerung in Mio.
* `emp`: Erwerbstätige in Mio.
* `hc`: Humankapitalindex (basierend auf Schuljahren und Bildungsrenditen)

`head()` zeigt die ersten fünf Zeilen. Die Zahlen links (630 bis 634) sind der Index aus dem ursprünglichen Datensatz, in dem Österreich nicht in der ersten Zeile steht.

`matplotlib`-Grafiken folgen einer Art Baukastenprinzip, wobei jeder Baukasten mit dem Befehl `plt.figure()` beginnt. Damit wird ein leeres Grafik-Objekt erstellt. Noch genauer: Der Befehl besagt lediglich, *dass* eine Grafik erstellt werden soll.

```python
# Erstelle eine leere Grafik
fig = plt.figure()
```

Die Ausgabe `<Figure size 640x480 with 0 Axes>` bestätigt, dass eine Grafik der Größe 640 × 480 Pixel angelegt wurde, die aber noch keine Achsen und damit keinen Inhalt hat.

Mit dem folgenden Befehl wird schließlich ein leeres Grafik-Objekt innerhalb von `fig` erstellt:

```python
# Füge der Grafik ein Koordinatensystem (Achsen) hinzu
ax = fig.add_subplot()
```

Diese Zelle erzeugt keine Ausgabe, weil das Ergebnis nur in der Variable `ax` gespeichert wird. Die Grafik selbst wurde in der vorherigen Zelle bereits angezeigt und wird hier nicht erneut dargestellt.

*NB: Mit dem Befehl `fig.add_subplot(2, 3, 1)` würden 2 x 3 leere Grafikobjekte erzeugt werden, die separat befüllt werden können. Das ist allerdings ein Thema für einen Folgekurs.*

Mit dem Befehl `ax.plot` kann das leere Grafikobjekt mit Inhalt befüllt werden.

```python
# Zeichne das reale BIP (rgdpe) über die Jahre als Linie
ax.plot(pwt_at.loc[:, "year"], pwt_at.loc[:, "rgdpe"])
```

Die Ausgabe ist nur eine Beschreibung des erzeugten Linienobjekts, aber keine Grafik. Jupyter zeigt eine Grafik nur am Ende der Zelle an, in der sie erstellt wurde. Da `fig` in einer früheren Zelle erstellt wurde, erscheint hier nichts.

Aus technischen Gründen müssen die drei beschriebenen Zeilen in Jupyter in einer Notebook-Zelle ausgeführt werden:

```python
# Erstelle Grafik und Achsen und zeichne die Linie, alles in einer Zelle
fig = plt.figure()
ax = fig.add_subplot()
ax.plot(pwt_at.loc[:, "year"], pwt_at.loc[:, "rgdpe"])
```

Jetzt erscheint die Grafik: Das reale BIP Österreichs ist von 1950 bis 2019 (dem letzten Jahr im Datensatz) auf mehr als das Zehnfache gewachsen. Phasen der Stagnation bzw. kleine Einbrüche sind Anfang der 1980er-Jahre, um 2001 und um die Finanzkrise 2009 zu erkennen. Die Achsen haben noch keine Beschriftung, das ließe sich mit `ax.set_xlabel` und `ax.set_ylabel` ergänzen. Die Zeile `[<matplotlib.lines.Line2D …>]` über der Grafik ist wieder die Beschreibung der Linie. Sie verschwindet, wenn man `plt.show()` als letzte Zeile ergänzt oder die Zeile mit `;` beendet.
